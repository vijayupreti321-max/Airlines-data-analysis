##**Average Price by Airline, Class, Destination City & Source city...**

```python
grouped_price_by_source = df.groupby(["source_city", "airline", "class", "destination_city"])["price"].mean().unstack()
grouped_price_by_source
```

##**Visualizing Average Price by Airline, Class, Destination City & Source city...**

```python
for source_city in grouped_price_by_source.index.get_level_values('source_city').unique():
    plt.figure(figsize=(12, 5))
    sns.heatmap(grouped_price_by_source.loc[source_city], annot=True, fmt=".0f", cmap="YlGnBu", linewidths=0.5, linecolor="white")
    plt.title(f"Average Price from {source_city} by Airline, Class, and Destination City")
    plt.xlabel("Destination City")
    plt.ylabel("Airline and Class")
    plt.tight_layout()
    plt.show()
```

##**Maximum Price by Airline, Class, Destination City & Source city...**

```python
Maximum_price=df.groupby(["airline","class","source_city","destination_city"])["price"].max().unstack()
Maximum_price
```

##**Visualizing Maximum Price by Airline, Class, Destination City & Source city...**

```python
for (airline, class_val, source_city) in Maximum_price.index:
    data_to_plot = Maximum_price.loc[(airline, class_val, source_city)]
    if not data_to_plot.empty:
        plt.figure(figsize=(12,5))

        sns.heatmap(data_to_plot.to_frame().T, annot=True, fmt=".0f", cmap="YlGnBu", linewidths=0.5, linecolor="white")
        plt.title(f"Maximum Price from {source_city} for {airline} ({class_val}) by Destination City")
        plt.xlabel("Destination City")
        plt.ylabel(f"{airline} ({class_val})")
        plt.tight_layout()
        plt.show()
```

##**Minimum Price by Airline, Class, Destination City & Source city...**

```python
Min_price=df.groupby(["airline","class","destination_city","source_city"])["price"].min().unstack()
Min_price
```

##**Visualizing Minimum Price by Airline, Class, Destination City & Source city...**

```python
for (airline, class_val, source_city) in Min_price.index:
    data_to_plot = Min_price.loc[(airline, class_val, source_city)]
    if not data_to_plot.empty:
        plt.figure(figsize=(12,5))
        sns.heatmap(data_to_plot.to_frame().T, annot=True, fmt=".0f", cmap="YlGnBu", linewidths=0.5, linecolor="white")
        plt.title(f"Minimum Price for {airline} ({class_val}) to {source_city} by Source City")
        plt.xlabel("Source City") # Updated label to reflect the heatmap axis
        plt.ylabel(f"{airline} ({class_val}) to {source_city}") # Updated ylabel for clarity
        plt.tight_layout()
        plt.show()
```

##**Correlation between "Duration" , "Days_left" & "Price"...**

```python
numerical_df = df[['duration', 'days_left', 'price']]

correlation_matrix = numerical_df.corr()

display(correlation_matrix)
```

##**Visualizing Correlation between "Duration" , "Days_left" & "Price"...**

```python
# Create a heatmap of the correlation matrix
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", linewidths=.5)
plt.title("Correlation Matrix of Numerical Columns")
plt.show()
```
